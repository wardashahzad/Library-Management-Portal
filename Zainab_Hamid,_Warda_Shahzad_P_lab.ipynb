# 1. Importing Required Libraries
In this section, we import the JSON library to store and manage book records in a structured format. We also define the file name that will be used to save library data.



```python
import json
import os

FILE_NAME = "books.json"
```

# 2. File Handling Functions
These functions will load existing book records from the JSON file and save updated records whenever changes are made.




def load_records():
    try:
        with open(FILE_NAME, "r") as file:
            books = json.load(file)
            return books
    except FileNotFoundError:
        return []
    except json.JSONDecodeError:
        print("Error: File contains invalid data.")
        return []

def save_records(books):
    try:
        with open(FILE_NAME, "w") as file:
            json.dump(books, file, indent=4)
        print("Records saved successfully.")
    except PermissionError:
        print("Error: Permission denied.")
    except OSError:
        print("Error: Unable to save records.")


```python

books = load_records()
print(books)
```

# Add a New Book

```python
def add_record(books):
    book_id = input("Enter Book ID: ")
    title = input("Enter Book Title: ")
    author = input("Enter Author Name: ")
    category = input("Enter Category: ")
    quantity = int(input("Enter Quantity: "))

    book = {
        "book_id": book_id,
        "title": title,
        "author": author,
        "category": category,
        "quantity": quantity,
        "available": quantity,
        "issued_to": []
    }

    books.append(book)
    save_records(books)
```

# Search a book

```python
def search_record(books):
    search_id = input("Enter Book ID: ")

    for book in books:
        if book["book_id"] == search_id:
            print(book)
            return book

    print("Book not found.")
    return None
```

# Display all books

```python
def display_records(books):
    if len(books) == 0:
        print("No books available.")
    else:
        for book in books:
            print(book)
```

# Issue book

```python
def issue_book(books):
    book_id = input("Enter Book ID: ")

    for book in books:
        if book["book_id"] == book_id:
            if book["available"] > 0:
                name = input("Enter Student Name: ")
                roll_no = input("Enter Roll Number: ")

                student = {"name": name, "roll_no": roll_no}
                book["issued_to"].append(student)

                book["available"] -= 1
                save_records(books)
                print("Book issued successfully.")
                return

            else:
                print("Book is not available.")
                return

    print("Book not found.")
```

# Return book

```python
def return_book(books):
    book_id = input("Enter Book ID: ")
    roll_no = input("Enter Student Roll Number: ")

    for book in books:
        if book["book_id"] == book_id:
            for student in book["issued_to"]:
                if student["roll_no"] == roll_no:
                    book["issued_to"].remove(student)
                    book["available"] += 1
                    save_records(books)
                    print("Book returned successfully.")
                    return

            print("Student record not found.")
            return

    print("Book not found.")
```

# Generate library report

```python
def generate_report(books):
    total = 0
    available = 0
    issued = 0

    for book in books:
        total += book["quantity"]
        available += book["available"]
        issued += len(book["issued_to"])

    print("Total Books:", total)
    print("Available Books:", available)
    print("Issued Books:", issued)
```

# Generate report

```python
def generate_report(books):
    total = 0
    available = 0

    for book in books:
        total += book["quantity"]
        available += book["available"]

    print("Total Books:", total)
    print("Available Books:", available)
    print("Issued Books:", total - available)
```

# User login

```python
def user_login():
    username = input("Enter Username: ")
    password = input("Enter Password: ")

    if username == "admin" and password == "1234":
        print("Login Successful!")
        return True
    else:
        print("Invalid Login!")
        return False
```

# Main menu

```python
def main():
    books = load_records()

    if user_login():
        while True:
            print("\nLIBRARY MANAGEMENT PORTAL")
            print("1. Add Book")
            print("2. Search Book")
            print("3. Display Books")
            print("4. Issue Book")
            print("5. Return Book")
            print("6. Generate Report")
            print("7. Exit")

            choice = input("Enter your choice: ")

            if choice == "1":
                add_record(books)
            elif choice == "2":
                search_record(books)
            elif choice == "3":
                display_records(books)
            elif choice == "4":
                issue_book(books)
            elif choice == "5":
                return_book(books)
            elif choice == "6":
                generate_report(books)
            elif choice == "7":
                print("Thank you!")
                break
            else:
                print("Invalid choice.")
```

```python
main()
```
